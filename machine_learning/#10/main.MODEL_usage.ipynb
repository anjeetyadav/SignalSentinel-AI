```python
import joblib
import pandas as pd

# 1. Load your saved model
model = joblib.load('SignalSentinel.joblib')
print("Model loaded successfully!")
print(f"ℹ️ This model expects exactly {model.n_features_in_} features.")

# 2. Get user input
prompt_msg = f"Enter {model.n_features_in_} values separated by commas: "
user_response = __builtins__.input(prompt_msg)

try:
    # 3. Convert input string into a list of floats
    features = [float(x.strip()) for x in user_response.split(',')]

    # 4. Validate feature count
    if len(features) != model.n_features_in_:
        print(f"\n❌ Error: You entered {len(features)} values, but the model needs {model.n_features_in_}.")
    else:
        # 5. Wrap in a Pandas DataFrame using the model's original feature names (if available)
        if hasattr(model, "feature_names_in_"):
            input_data = pd.DataFrame([features], columns=model.feature_names_in_)
        else:
            input_data = [features]

        # 6. Make prediction
        prediction = model.predict(input_data)
        print(f"\n🎉 Predicted Result: {prediction[0]}")

except ValueError:
    print("\n❌ Error: Please make sure you only enter valid numbers separated by commas.")
```
